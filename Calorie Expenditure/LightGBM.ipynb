```python
import polars as pl
import numpy as np
import lightgbm as lgb
```

```python
def feature_engineer(df: pl.DataFrame, is_train: bool = True) -> pl.DataFrame:
    """
    Applies all feature engineering steps.
    `is_train` flag to handle target variable processing.
    """
    df_transformed = df.clone() # Work on a copy

    if is_train:
        # Target Transformation (only for training data)
        df_transformed = df_transformed.with_columns(
            pl.col("Calories").log1p().alias("Calories_log1p")
        )

    # Encoding Categorical Features ('Sex') - One-Hot Encoding
    df_transformed = df_transformed.with_columns([
        pl.col("Sex").eq(pl.lit("male")).cast(pl.UInt8).alias("Sex_male"),
        pl.col("Sex").eq(pl.lit("female")).cast(pl.UInt8).alias("Sex_female"),
    ])

    # Interaction Features
    # Some interactions might need the Sex_male column if you defined them that way.
    # For simplicity, ensure Sex_male/female exist before these are calculated
    # Or use pl.when() as in the original example.
    df_transformed = df_transformed.with_columns([
        (pl.col("Duration") * pl.col("Heart_Rate")).alias("Dur_HR_Interact"),
        (pl.col("Weight") * pl.col("Duration")).alias("Wt_Dur_Interact"),
        (pl.col("Weight") * pl.col("Heart_Rate")).alias("Wt_HR_Interact"),
        (pl.col("Age") * pl.col("Heart_Rate")).alias("Age_HR_Interact"),
        (pl.col("Body_Temp") * pl.col("Duration")).alias("Temp_Dur_Interact"),
        (pl.col("Sex_male") * pl.col("Duration")).alias("SexMale_Dur_Interact"),
        (pl.col("Sex_male") * pl.col("Weight")).alias("SexMale_Wt_Interact"),
    ])

    # Polynomial Features (Degree 2 examples)
    df_transformed = df_transformed.with_columns([
        pl.col("Age").pow(2).alias("Age_sq"),
        pl.col("Duration").pow(2).alias("Duration_sq"),
        pl.col("Heart_Rate").pow(2).alias("Heart_Rate_sq"),
    ])

    # Ratio Features / Derived Metrics
    df_transformed = df_transformed.with_columns([
        (pl.col("Weight") / (pl.col("Height") / 100.0).pow(2)).alias("BMI"),
        (pl.col("Heart_Rate") / (220.0 - pl.col("Age")).fill_nan(0).replace(np.inf, 0)).alias("HR_perc_Max_Approx"), # Handle potential division by zero or Age >= 220
    ])

    # Binning/Grouping
    # For XGBoost, you might use enable_categorical=True or one-hot encode it.
    # Here, I'll one-hot encode it for general compatibility.
    df_age_groups = df_transformed.with_columns(
        pl.when(pl.col("Age") <= 30).then(pl.lit("Age_lte_30"))
        .when((pl.col("Age") > 30) & (pl.col("Age") <= 40)).then(pl.lit("Age_31_40"))
        .when((pl.col("Age") > 40) & (pl.col("Age") <= 50)).then(pl.lit("Age_41_50"))
        .when((pl.col("Age") > 50) & (pl.col("Age") <= 60)).then(pl.lit("Age_51_60"))
        .otherwise(pl.lit("Age_gt_60"))
        .alias("Age_Group_Cat")
    ).to_dummies(columns=["Age_Group_Cat"]) # One-hot encode the new categorical age group

    # Polars to_dummies might create column names like "Age_Group_Cat_Age_lte_30"
    # We need to merge these back. The original df_transformed doesn't have these yet.
    # A safer way is to join if to_dummies returns a new df with only dummy columns + keys
    # Or, more simply for this case, create them directly in df_transformed

    # Re-doing Age Grouping and One-Hot Encoding directly for clarity
    age_group_exprs = [
        (pl.col("Age") <= 30).cast(pl.UInt8).alias("Age_Group_lte_30"),
        ((pl.col("Age") > 30) & (pl.col("Age") <= 40)).cast(pl.UInt8).alias("Age_Group_31_40"),
        ((pl.col("Age") > 40) & (pl.col("Age") <= 50)).cast(pl.UInt8).alias("Age_Group_41_50"),
        ((pl.col("Age") > 50) & (pl.col("Age") <= 60)).cast(pl.UInt8).alias("Age_Group_51_60"),
        (pl.col("Age") > 60).cast(pl.UInt8).alias("Age_Group_gt_60"),
    ]
    df_transformed = df_transformed.with_columns(age_group_exprs)

    return df_transformed
```

```python
feature_cols = [
    'Age', 'Height', 'Weight', 'Duration', 'Heart_Rate', 'Body_Temp', # Original numerics
    'Sex_male', 'Sex_female',             # Encoded Sex
    'Dur_HR_Interact', 'Wt_Dur_Interact', 'Wt_HR_Interact', 'Age_HR_Interact', # Interactions
    'Temp_Dur_Interact', 'SexMale_Dur_Interact', 'SexMale_Wt_Interact',
    'Age_sq', 'Duration_sq', 'Heart_Rate_sq', # Polynomials
    'BMI', 'HR_perc_Max_Approx',            # Ratios
    'Age_Group_lte_30', 'Age_Group_31_40', 'Age_Group_41_50', # Age Group Dummies
    'Age_Group_51_60', 'Age_Group_gt_60'
]
target_col_log = 'Calories_log1p'
```

```python
print("Loading and processing training data...")
train_df_raw = pl.read_csv("data/train.csv")
# Ensure 'id' is Int64 if not automatically detected
train_df_raw = train_df_raw.with_columns(pl.col("id").cast(pl.Int64))
train_df_processed = feature_engineer(train_df_raw, is_train=True)

X_train = train_df_processed.select(feature_cols).to_numpy()
y_train_log = train_df_processed.select(target_col_log).to_numpy().ravel()
```

```python
final_cols_to_select = list(set(feature_cols + [target_col_log])) # Use set to avoid duplicates

pandas_df = train_df_processed.select(final_cols_to_select).to_pandas()

# Separate features (X) and target (y) into pandas objects
X_pd = pandas_df[feature_cols]
y_pd = pandas_df[target_col_log] # This is already log1p transformed

print(f"\nShape of X_pd: {X_pd.shape}")
print(f"Shape of y_pd: {y_pd.shape}")
print("\nPandas X head:")
print(X_pd.head())
print("\nPandas y head:")
print(y_pd.head())
```

```python
from optuna.pruners import MedianPruner
from optuna.integration import XGBoostPruningCallback
```

```python
import optuna
import lightgbm as lgb # Import LightGBM
from optuna.integration import LightGBMPruningCallback # LightGBM specific pruning
from sklearn.model_selection import KFold
from sklearn.metrics import mean_squared_error
from category_encoders import TargetEncoder
import numpy as np
import pandas as pd

def objective_lgbm(trial, X, y): # Add X, y as arguments
    # Parameters for LightGBM
    # Note: LightGBM often performs well with more leaves and not too deep trees.
    # 'max_depth' can sometimes be less critical than 'num_leaves'.
    optuna_params_lgbm = {
        'n_estimators': trial.suggest_int('n_estimators', 500, 5000, step=500),
        'learning_rate': trial.suggest_float('learning_rate', 1e-3, 0.3, log=True),
        'num_leaves': trial.suggest_int('num_leaves', 20, 300), # Key parameter for LightGBM
        'max_depth': trial.suggest_int('max_depth', 3, 12), # Can be -1 (no limit) if num_leaves is well-tuned
        'subsample': trial.suggest_float('subsample', 0.5, 1.0), # (bagging_fraction)
        'colsample_bytree': trial.suggest_float('colsample_bytree', 0.5, 1.0), # (feature_fraction)
        'reg_alpha': trial.suggest_float('reg_alpha', 1e-8, 1.0, log=True), # L1 regularization
        'reg_lambda': trial.suggest_float('reg_lambda', 1e-8, 1.0, log=True), # L2 regularization
        'min_child_samples': trial.suggest_int('min_child_samples', 5, 100), # Min data in a leaf
        'min_split_gain': trial.suggest_float('min_split_gain', 1e-8, 1.0, log=True), # Like XGBoost's gamma
        # 'boosting_type': trial.suggest_categorical('boosting_type', ['gbdt', 'dart', 'goss']), # gbdt is default and often best
    }

    # Constructing the params dictionary for LGBMRegressor
    # Many Optuna suggested params map directly to LGBMRegressor arguments
    params = {
        'objective': 'regression_l1', # Or 'regression' for RMSE. MAE (regression_l1) can be robust.
                                      # 'rmse' is also a valid objective. For RMSLE on original scale,
                                      # training on log(target) with RMSE objective is common.
                                      # Let's stick to 'rmse' objective for consistency with your XGBoost eval_metric.
        'objective': 'rmse',
        'metric': 'rmse',             # Metric reported during training
        'n_estimators': optuna_params_lgbm['n_estimators'],
        'learning_rate': optuna_params_lgbm['learning_rate'],
        'num_leaves': optuna_params_lgbm['num_leaves'],
        'max_depth': optuna_params_lgbm['max_depth'],
        'subsample': optuna_params_lgbm['subsample'],
        'colsample_bytree': optuna_params_lgbm['colsample_bytree'],
        'reg_alpha': optuna_params_lgbm['reg_alpha'],
        'reg_lambda': optuna_params_lgbm['reg_lambda'],
        'min_child_samples': optuna_params_lgbm['min_child_samples'],
        'min_split_gain': optuna_params_lgbm['min_split_gain'],
        'random_state': 42,
        'n_jobs': -1,
        'verbose': -1, # Suppress LightGBM's own verbosity during Optuna trials
        # 'boosting_type': optuna_params_lgbm.get('boosting_type', 'gbdt') # if you add boosting_type to optuna_params
    }

    kf = KFold(n_splits=5, shuffle=True, random_state=42)
    rmse_scores = []

    # Identify categorical columns for TargetEncoder (same as your XGBoost setup)
    categorical_cols_for_encoder = X.select_dtypes(include=['object', 'category']).columns.tolist()

    for fold, (train_idx, valid_idx) in enumerate(kf.split(X, y)):
        X_train, X_valid = X.iloc[train_idx].copy(), X.iloc[valid_idx].copy()
        y_train, y_valid = y.iloc[train_idx], y.iloc[valid_idx]

        if categorical_cols_for_encoder:
            # Use the same smoothing hyperparameter as in your XGBoost setup or define a new one
            smoothing_lgbm = trial.suggest_float("smoothing_lgbm", 0.05, 20.0, log=True) # Potentially different range or name
            enc = TargetEncoder(cols=categorical_cols_for_encoder, smoothing=smoothing_lgbm, handle_unknown='value', handle_missing='value')
            enc.fit(X_train, y_train)
            X_train = enc.transform(X_train)
            X_valid = enc.transform(X_valid)

        # LightGBM can often handle column names with special characters,
        # but ensuring they are strings is good practice.
        # TargetEncoder should produce numerical columns anyway for the encoded ones.
        X_train.columns = X_train.columns.astype(str)
        X_valid.columns = X_valid.columns.astype(str)

        model = lgb.LGBMRegressor(**params)

        # Pruning callback for LightGBM
        # It monitors the validation metric. The metric name must match what LightGBM reports.
        # When using LGBMRegressor and eval_set, the default validation metric name is 'valid_0-<metric_name>'
        pruning_callback = LightGBMPruningCallback(trial, "rmse", valid_name="valid_0")

        model.fit(
            X_train, y_train,
            eval_set=[(X_valid, y_valid)],
            eval_metric="rmse", # This is the metric it will report and pruning callback will use
            callbacks=[
                #lgb.early_stopping(rounds=50, verbose=False), # LightGBM's early stopping
                pruning_callback
            ],
            # verbose=-1 # Already set in params, but can be set here too
        )

        preds = model.predict(X_valid)
        rmse = np.sqrt(mean_squared_error(y_valid, preds))
        rmse_scores.append(rmse)

    mean_rmse = np.mean(rmse_scores)
    return mean_rmse
```

```python
study = optuna.create_study(direction='minimize', pruner=MedianPruner(n_warmup_steps=5))
study.optimize(lambda trial: objective_lgbm(trial, X_pd, y_pd), n_trials=40)

print("Best CV RMSE:", study.best_value)
print("Best Hyperparameters:", study.best_params)
```

```python
best_params = study.best_params
model_lgb = lgb.LGBMRegressor(**best_params)
model_lgb.fit(X_train, y_train_log)
```

```python
print("Loading test data...")
test_df_raw = pl.read_csv("data/test.csv")
test_df_raw = test_df_raw.with_columns(pl.col("id").cast(pl.Int64))
test_ids = test_df_raw.select("id")

# --- Apply Feature Engineering to Test Data ---
print("Processing test data for LightGBM...")
test_df_processed = feature_engineer(test_df_raw, is_train=False)
X_test = test_df_processed.select(feature_cols).to_numpy()
```

```python
print("Making LightGBM predictions...")
predictions_log_lgb = model_lgb.predict(X_test)
```

```python
predictions_calories_lgb = np.expm1(predictions_log_lgb)
predictions_calories_lgb = np.maximum(0, predictions_calories_lgb)
```

```python
print("Creating LightGBM submission file...")
submission_df_lgb = test_ids.with_columns(
    pl.Series("Calories", predictions_calories_lgb)
)

submission_df_lgb.write_csv("submissions/lgbm.csv")
print("Submission file 'lgbm.csv' created successfully.")
print(submission_df_lgb.head())
```
